# Stage 01 — Infinite-width Gaussian-RBF periodic advection predictor

Python/Jupyter conversion of `STAGE_01_INF_RBF_ADV_PAPER.m`, preserving the periodic boundary kernel, slab marching, independent residual probes, and periodic low/mid/high attention buffering.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import eigh
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erf
from pathlib import Path
from time import perf_counter

def radical_inverse(n, b):
    n = np.asarray(n, dtype=int).copy()
    r = np.zeros_like(n, dtype=float)
    f = 1.0 / b
    while np.any(n > 0):
        r += f * (n % b)
        n //= b
        f /= b
    return r

def halton_2d_prefix(N):
    idx = np.arange(1, N+1)
    return radical_inverse(idx, 2), radical_inverse(idx, 3)

def dilate_periodic(mask, radius):
    mask = np.asarray(mask, dtype=bool)
    N = mask.size
    out = np.zeros_like(mask)
    for j in np.flatnonzero(mask):
        for r in range(-radius, radius+1):
            out[(j+r) % N] = True
    return out

def attention_to_levels(score, theta_mid, theta_high, high_radius, mid_radius):
    score = np.asarray(score, dtype=float).ravel()
    if np.max(score) <= np.finfo(float).eps:
        return np.ones(score.size, dtype=int)
    high_raw = score >= theta_high
    mid_raw = score >= theta_mid
    if not np.any(high_raw):
        high_raw[np.argmax(score)] = True
    high_mask = dilate_periodic(high_raw, high_radius)
    mid_mask = dilate_periodic(mid_raw | high_raw, mid_radius)
    levels = np.ones(score.size, dtype=int)
    levels[mid_mask] = 2
    levels[high_mask] = 3
    return levels

def rbf1d_uniform_derivs(x, y, sigma, a, b):
    x = np.asarray(x, dtype=float).reshape(-1, 1)
    y = np.asarray(y, dtype=float).reshape(1, -1)
    m = 0.5*(x+y)
    d = x-y
    za = a-m
    zb = b-m
    Ed = np.exp(-(d**2)/(4*sigma**2))
    ea = np.exp(-(za/sigma)**2)
    eb = np.exp(-(zb/sigma)**2)
    E = erf(zb/sigma)-erf(za/sigma)
    J0 = 0.5*sigma*np.sqrt(np.pi)*E
    J1 = 0.5*sigma**2*(ea-eb)
    J2 = 0.25*sigma**3*np.sqrt(np.pi)*E - 0.5*sigma**2*(zb*eb-za*ea)
    C = 1/(b-a)
    K = C*Ed*J0
    Ix = 0.5*d*J0 - J1
    Iy = -0.5*d*J0 - J1
    K1 = -C*Ed*Ix/sigma**2
    K2 = -C*Ed*Iy/sigma**2
    K12 = C*Ed*(J2 - 0.25*d**2*J0)/sigma**4
    return K, K1, K2, K12

def compute_kernel_norm(cfg):
    xr = 0.5*(cfg["XL"]+cfg["XR"])
    tr = 0.5*cfg["dt"]
    Kx = rbf1d_uniform_derivs([xr],[xr],cfg["sigma_x"],cfg["cx_a"],cfg["cx_b"])[0]
    Kt = rbf1d_uniform_derivs([tr],[tr],cfg["sigma_t"],cfg["ct_a"],cfg["ct_b"])[0]
    return float((Kx*Kt)[0,0])

def kernel_2d(cfg, x1,t1,x2,t2,type1,type2):
    Kx,Kx1,Kx2,Kx12 = rbf1d_uniform_derivs(x1,x2,cfg["sigma_x"],cfg["cx_a"],cfg["cx_b"])
    Kt,Kt1,Kt2,Kt12 = rbf1d_uniform_derivs(t1,t2,cfg["sigma_t"],cfg["ct_a"],cfg["ct_b"])
    if type1=="val" and type2=="val":
        K = Kx*Kt
    elif type1=="pde" and type2=="val":
        K = Kx1*Kt + Kx*Kt1
    elif type1=="val" and type2=="pde":
        K = Kx2*Kt + Kx*Kt2
    elif type1=="pde" and type2=="pde":
        K = Kx12*Kt + Kx1*Kt2 + Kx2*Kt1 + Kx*Kt12
    else:
        raise ValueError("Unknown kernel functional type.")
    return K/cfg["norm_const"]

def kernel_val_vs_bc(cfg,x,t,tB):
    tB=np.asarray(tB)
    return (kernel_2d(cfg,x,t,np.full_like(tB,cfg["XL"],dtype=float),tB,"val","val")
            - kernel_2d(cfg,x,t,np.full_like(tB,cfg["XR"],dtype=float),tB,"val","val"))

def kernel_pde_vs_bc(cfg,x,t,tB):
    tB=np.asarray(tB)
    return (kernel_2d(cfg,x,t,np.full_like(tB,cfg["XL"],dtype=float),tB,"pde","val")
            - kernel_2d(cfg,x,t,np.full_like(tB,cfg["XR"],dtype=float),tB,"pde","val"))

def kernel_bc_vs_val(cfg,tB,x,t):
    tB=np.asarray(tB)
    return (kernel_2d(cfg,np.full_like(tB,cfg["XL"],dtype=float),tB,x,t,"val","val")
            - kernel_2d(cfg,np.full_like(tB,cfg["XR"],dtype=float),tB,x,t,"val","val"))

def kernel_bc_vs_bc(cfg,t1,t2):
    t1=np.asarray(t1); t2=np.asarray(t2)
    XL1=np.full_like(t1,cfg["XL"],dtype=float); XR1=np.full_like(t1,cfg["XR"],dtype=float)
    XL2=np.full_like(t2,cfg["XL"],dtype=float); XR2=np.full_like(t2,cfg["XR"],dtype=float)
    return (kernel_2d(cfg,XL1,t1,XL2,t2,"val","val")
            - kernel_2d(cfg,XL1,t1,XR2,t2,"val","val")
            - kernel_2d(cfg,XR1,t1,XL2,t2,"val","val")
            + kernel_2d(cfg,XR1,t1,XR2,t2,"val","val"))

def build_constraint_gram(cfg,tr):
    xP,tP,tB,xI,tI = tr["xP"],tr["tP"],tr["tB"],tr["xI"],tr["tI"]
    wP,wB,wI = tr["wP"],tr["wB"],tr["wI"]
    KPP=kernel_2d(cfg,xP,tP,xP,tP,"pde","pde")
    KPB=kernel_pde_vs_bc(cfg,xP,tP,tB)
    KPI=kernel_2d(cfg,xP,tP,xI,tI,"pde","val")
    KBB=kernel_bc_vs_bc(cfg,tB,tB)
    KBI=kernel_bc_vs_val(cfg,tB,xI,tI)
    KII=kernel_2d(cfg,xI,tI,xI,tI,"val","val")
    K=np.block([
        [wP*wP*KPP, wP*wB*KPB, wP*wI*KPI],
        [wB*wP*KPB.T, wB*wB*KBB, wB*wI*KBI],
        [wI*wP*KPI.T, wI*wB*KBI.T, wI*wI*KII],
    ])
    return 0.5*(K+K.T)

def predict_value(cfg,tr,a,xq,tq):
    NP=len(tr["xP"]); NB=len(tr["tB"])
    aP=a[:NP]; aB=a[NP:NP+NB]; aI=a[NP+NB:]
    KP=kernel_2d(cfg,xq,tq,tr["xP"],tr["tP"],"val","pde")
    KB=kernel_val_vs_bc(cfg,xq,tq,tr["tB"])
    KI=kernel_2d(cfg,xq,tq,tr["xI"],tr["tI"],"val","val")
    return (tr["wP"]*KP)@aP + (tr["wB"]*KB)@aB + (tr["wI"]*KI)@aI

def predict_pde_residual(cfg,tr,a,xq,tq):
    NP=len(tr["xP"]); NB=len(tr["tB"])
    aP=a[:NP]; aB=a[NP:NP+NB]; aI=a[NP+NB:]
    KP=kernel_2d(cfg,xq,tq,tr["xP"],tr["tP"],"pde","pde")
    KB=kernel_pde_vs_bc(cfg,xq,tq,tr["tB"])
    KI=kernel_2d(cfg,xq,tq,tr["xI"],tr["tI"],"pde","val")
    return (tr["wP"]*KP)@aP + (tr["wB"]*KB)@aB + (tr["wI"]*KI)@aI
```

```python
XL, XR = -1.0, 1.0
L = XR - XL
nSlabs, dt_slab = 100, 0.01
Tfinal = nSlabs*dt_slab
pulse_beta = 25.0

def init(x):
    x=np.asarray(x,float)
    return np.exp(-(pulse_beta*x)**2)

def uexact(x,t):
    x=np.asarray(x,float)
    return init(np.mod(x - t - XL, L) + XL)

NBx=25
x_edges=np.linspace(XL,XR,NBx+1)
Npde_per_block_pilot=8
Nic=401
Nbc=11
probe_nx,probe_nt=10,7
attention_rms_weight,attention_max_weight=0.70,0.30
theta_high,theta_mid=0.35,0.10
high_buffer_radius_x,mid_buffer_radius_x=1,2
sigma_x,sigma_t=0.020,0.010
cx_a,cx_b=XL-0.20*L, XR+0.20*L
ct_a,ct_b=-1.5*dt_slab, 2.5*dt_slab
ridge_rel,eig_tol_rel=1e-9,1e-13

hx01,ht01=halton_2d_prefix(Npde_per_block_pilot)
xP=np.zeros(NBx*Npde_per_block_pilot); tP=np.zeros_like(xP)
kk=0
for jx in range(NBx):
    xa,xb=x_edges[jx],x_edges[jx+1]
    sl=slice(kk,kk+Npde_per_block_pilot)
    xP[sl]=xa+(xb-xa)*hx01
    tP[sl]=dt_slab*ht01
    kk+=Npde_per_block_pilot

xI=np.linspace(XL,XR,Nic)
tI=np.zeros(Nic)
tB=np.linspace(0,dt_slab,Nbc)

Npde=len(xP)
wP,wB,wI=1/np.sqrt(Npde),1/np.sqrt(Nbc),1/np.sqrt(Nic)

kcfg=dict(sigma_x=sigma_x,sigma_t=sigma_t,cx_a=cx_a,cx_b=cx_b,
          ct_a=ct_a,ct_b=ct_b,XL=XL,XR=XR,dt=dt_slab)
kcfg["norm_const"]=compute_kernel_norm(kcfg)

train=dict(xP=xP,tP=tP,tB=tB,xI=xI,tI=tI,wP=wP,wB=wB,wI=wI)
print("STAGE 01: INFINITE-RBF ADVECTION PREDICTOR")
print(f"pulse beta={pulse_beta:.1f}, pilot PDE points={Npde}, sigma=({sigma_x:.3f},{sigma_t:.3f})")
```

```python
_t_stage1_total0 = perf_counter()

_t0 = perf_counter()
Ktrain = build_constraint_gram(kcfg, train)
t_stage1_assembly = perf_counter() - _t0

_t0 = perf_counter()
d, V = eigh(Ktrain)
dmax = np.max(d)
if dmax <= 0:
    raise RuntimeError("Predictor kernel Gram matrix is not positive on the retained scale.")
lam = ridge_rel * dmax
keep = d > eig_tol_rel * dmax
Vin = V[:, keep]
din = d[keep] + lam
t_stage1_factorization = perf_counter() - _t0

print(f"constraints={Ktrain.shape[0]}, retained={np.sum(keep)}/{len(d)}, lambda={lam:.3e}")
print(f"Kernel/operator assembly time = {t_stage1_assembly:.6f} s")
print(f"Spectral factorization time   = {t_stage1_factorization:.6f} s")
```

```python
uI=init(xI)
Nx_plot=801
x_plot=np.linspace(XL,XR,Nx_plot)
plot_times=np.arange(nSlabs+1)*dt_slab
Uplot=np.zeros((nSlabs+1,Nx_plot))
Uplot[0]=init(x_plot)

Rmat=np.zeros((nSlabs,NBx))
Rmax_mat=np.zeros((nSlabs,NBx))
attention_score_mat=np.zeros((nSlabs,NBx))
Lmat=np.zeros((nSlabs,NBx),dtype=int)
coeff_all=np.zeros((Ktrain.shape[0],nSlabs))

t_stage1_coeff = 0.0
t_stage1_residual_probe = 0.0
t_stage1_prediction = 0.0

for s in range(nSlabs):
    TL=s*dt_slab
    rhs=np.r_[np.zeros(Npde),np.zeros(Nbc),wI*uI]

    _t0 = perf_counter()
    coeff=Vin@((Vin.T@rhs)/din)
    t_stage1_coeff += perf_counter() - _t0
    coeff_all[:,s]=coeff

    _t0 = perf_counter()
    Rrms=np.zeros(NBx); Rmx=np.zeros(NBx)
    for jx in range(NBx):
        xa,xb=x_edges[jx],x_edges[jx+1]
        xp=np.linspace(xa,xb,probe_nx+2)[1:-1]
        tp=np.linspace(0,dt_slab,probe_nt+2)[1:-1]
        XP,TP=np.meshgrid(xp,tp)
        rq=predict_pde_residual(kcfg,train,coeff,XP.ravel(),TP.ravel())
        Rrms[jx]=np.sqrt(np.mean(rq**2))
        Rmx[jx]=np.max(np.abs(rq))

    nr=Rrms/max(np.max(Rrms),np.finfo(float).eps)
    nm=Rmx/max(np.max(Rmx),np.finfo(float).eps)
    score=attention_rms_weight*nr+attention_max_weight*nm
    lev=attention_to_levels(score,theta_mid,theta_high,high_buffer_radius_x,mid_buffer_radius_x)
    t_stage1_residual_probe += perf_counter() - _t0

    Rmat[s]=Rrms
    Rmax_mat[s]=Rmx
    attention_score_mat[s]=score
    Lmat[s]=lev

    _t0 = perf_counter()
    uq_top=predict_value(kcfg,train,coeff,x_plot,np.full_like(x_plot,dt_slab))
    Uplot[s+1]=uq_top
    uI=predict_value(kcfg,train,coeff,xI,np.full_like(xI,dt_slab))
    t_stage1_prediction += perf_counter() - _t0

u_ex_final=uexact(x_plot,Tfinal)
err_final=Uplot[-1]-u_ex_final
predictor_rmse=np.sqrt(np.mean(err_final**2))
predictor_max_error=np.max(np.abs(err_final))

t_stage1_core = t_stage1_assembly + t_stage1_factorization + t_stage1_coeff
t_stage1_total = perf_counter() - _t_stage1_total0

print(f"Predictor final RMSE={predictor_rmse:.3e}")
print(f"Predictor final max error={predictor_max_error:.3e}")
print(f"Coefficient solves (all slabs)= {t_stage1_coeff:.6f} s")
print(f"Residual probing/adaptation   = {t_stage1_residual_probe:.6f} s")
print(f"Slab prediction/transfer      = {t_stage1_prediction:.6f} s")
print(f"Predictor core CFKS time      = {t_stage1_core:.6f} s")
print(f"Stage-1 total elapsed time    = {t_stage1_total:.6f} s")
```

```python
np.savez_compressed(
    "STAGE_01_INF_RBF_ADV_attention_PAPER.npz",
    XL=XL,XR=XR,L=L,pulse_beta=pulse_beta,NBx=NBx,x_edges=x_edges,
    nSlabs=nSlabs,dt_slab=dt_slab,Tfinal=Tfinal,
    Npde_per_block_pilot=Npde_per_block_pilot,Npde=Npde,Nic=Nic,Nbc=Nbc,
    xP=xP,tP=tP,xI=xI,tI=tI,tB=tB,
    sigma_x=sigma_x,sigma_t=sigma_t,cx_a=cx_a,cx_b=cx_b,ct_a=ct_a,ct_b=ct_b,
    ridge_rel=ridge_rel,eig_tol_rel=eig_tol_rel,lam=lam,keep=keep,
    theta_high=theta_high,theta_mid=theta_mid,
    high_buffer_radius_x=high_buffer_radius_x,mid_buffer_radius_x=mid_buffer_radius_x,
    attention_rms_weight=attention_rms_weight,attention_max_weight=attention_max_weight,
    probe_nx=probe_nx,probe_nt=probe_nt,
    Rmat=attention_score_mat,Lmat=Lmat,Rrms_mat=Rmat,Rmax_mat=Rmax_mat,
    coeff_all=coeff_all,x_plot=x_plot,plot_times=plot_times,Uplot=Uplot,
    predictor_rmse=predictor_rmse,predictor_max_error=predictor_max_error,
    u_ex_final=u_ex_final,err_final=err_final,
    t_stage1_assembly=t_stage1_assembly,
    t_stage1_factorization=t_stage1_factorization,
    t_stage1_coeff=t_stage1_coeff,
    t_stage1_residual_probe=t_stage1_residual_probe,
    t_stage1_prediction=t_stage1_prediction,
    t_stage1_core=t_stage1_core,
    t_stage1_total=t_stage1_total
)
print("Saved STAGE_01_INF_RBF_ADV_attention_PAPER.npz")
```
